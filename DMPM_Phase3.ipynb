```python
import tkinter as tk
from tkinter import ttk, messagebox
import pandas as pd 
import numpy as np 
import nltk 
from nltk.corpus import stopwords 
from nltk.tokenize import word_tokenize 
from nltk.stem import WordNetLemmatizer 
from sklearn.feature_extraction.text import TfidfVectorizer 
from sklearn.naive_bayes import MultinomialNB 
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer 

# Download NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

# Load the dataset
data = pd.read_csv('Dataset-SA.csv')

# Data preprocessing
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    if isinstance(text, str):
        tokens = word_tokenize(text.lower())  # Tokenization and lowercase
        tokens = [token for token in tokens if token.isalnum()]  # Remove non-alphanumeric
        tokens = [token for token in tokens if token not in stop_words]  # Remove stopwords
        tokens = [lemmatizer.lemmatize(token) for token in tokens]  # Lemmatization
        return ' '.join(tokens)
    else:
        return ''

data['clean_review'] = data['Review'].apply(preprocess_text)

# Preprocess the 'Rate' column
data['Rate'] = data['Rate'].str.extract('(\d+)').astype(float)

# Feature extraction
tfidf = TfidfVectorizer(max_features=5000)
X = tfidf.fit_transform(data['clean_review'])
y = np.where(data['Rate'] >= 3, 'positive', 'negative')  # Assuming 'Rate' is the column for star ratings

# Sentiment analysis model training
classifier = MultinomialNB()
classifier.fit(X, y)

# Emotion detection using VADER
analyzer = SentimentIntensityAnalyzer()

def detect_emotion(text):
    scores = analyzer.polarity_scores(text)
    if scores['compound'] >= 0.05:
        return 'positive'
    elif scores['compound'] <= -0.05:
        return 'negative'
    else:
        return 'neutral'

# Define assess_product_quality function
def assess_product_quality(rating):
    if rating >= 4:
        return 'excellent'
    elif rating >= 3:
        return 'good'
    elif rating >= 2:
        return 'average'
    else:
        return 'poor'

def analyze_review(input_data):
    if isinstance(input_data, str):
        # Text input
        cleaned_text = preprocess_text(input_data)
        text_features = tfidf.transform([cleaned_text])
        sentiment = classifier.predict(text_features)[0]
        emotion = detect_emotion(input_data)
        return sentiment, emotion
    elif isinstance(input_data, (int, float)):
        # Numeric input
        product_quality = assess_product_quality(input_data)
        return product_quality
    else:
        return "Invalid input type"

def on_analyze():
    user_review = text_entry.get("1.0", tk.END).strip()
    user_rating = rating_entry.get().strip()  
    
    if not user_rating:
        messagebox.showerror("Error", "Please enter the product rating.")
        return
    
    try:
        user_rating = float(user_rating)  
    except ValueError:
        messagebox.showerror("Error", "Invalid product rating. Please enter a valid number.")
        return
    
    sentiment, emotion = analyze_review(user_review)
    product_quality = analyze_review(user_rating)
    
    sentiment_label.config(text=f"Sentiment: {sentiment}", foreground="blue")
    emotion_label.config(text=f"Emotion: {emotion}", foreground="green")
    quality_label.config(text=f"Product Quality: {product_quality}", foreground="orange")

# GUI setup
root = tk.Tk()
root.title("Product Review Analysis System")
root.configure(background="black")

# Text review entry
text_entry = tk.Text(root, height=5, width=50, bg="gray", fg="white", font=("Arial", 12))
text_entry.grid(row=0, column=0, padx=10, pady=10)

# Rating entry
rating_label = tk.Label(root, text="Product Rating (1-5):", bg="black", fg="white", font=("Arial", 12))
rating_label.grid(row=1, column=0, padx=10, pady=5, sticky="w")
rating_entry = tk.Entry(root, bg="gray", fg="white", font=("Arial", 12))
rating_entry.grid(row=1, column=0, padx=10, pady=5, sticky="e")

# Analyze button
analyze_button = tk.Button(root, text="Analyze", command=on_analyze, bg="green", fg="white", font=("Arial", 12), relief=tk.RAISED)
analyze_button.grid(row=1, column=1, padx=10, pady=5)

# Labels for sentiment, emotion, and product quality
sentiment_label = tk.Label(root, text="Sentiment:", bg="black", fg="white", font=("Arial", 12, "bold"))
sentiment_label.grid(row=2, column=0, padx=10, pady=5, sticky="w")
emotion_label = tk.Label(root, text="Emotion:", bg="black", fg="white", font=("Arial", 12, "bold"))
emotion_label.grid(row=3, column=0, padx=10, pady=5, sticky="w")
quality_label = tk.Label(root, text="Product Quality:", bg="black", fg="white", font=("Arial", 12, "bold"))
quality_label.grid(row=4, column=0, padx=10, pady=5, sticky="w")

root.mainloop()
```
